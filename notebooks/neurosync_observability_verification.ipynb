## 2. Forward Biophysics: Leadfields, Subspace Whitening, and Ambiguity Matrices
Here we execute the exact forward-model leadfield mapping for 10-20 channels and compute the rank-deficient whitening eigenspectrum and pairwise absolute cosine ambiguity $\rho_{ij} = |u_i^\top u_j|$.

```python
channels = ["Fz", "Cz", "Pz", "Oz", "CPz"]
G = get_canonical_p300_leadfield(channels)

# Simulate a high-impedance dry sensor covariance with rank-deficiency
rng = np.random.default_rng(42)
A = rng.standard_normal((len(channels), 3))
noise_cov = A @ A.T + np.diag([0.05, 0.05, 0.05, 0.05, 0.05])

# Evaluate Observable Geometry Certificate
report = compute_observability_certificate(G, noise_cov)
print(f"Source Observability Certificate (O_cert): {report.certificate:.4f}")
print(f"Worst-Case Forward Ambiguity (rho):       {report.worst_case_ambiguity:.4f}")
print(f"Supported Rank:                           {report.supported_rank}")
print(f"Resolution Fraction:                      {report.resolution_fraction:.4f}")
```
